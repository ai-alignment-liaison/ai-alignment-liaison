## Derive Product Requirements From A Single Value

### Functions

```python
# prompt: Use gpt-4 from openai to derive product requirements given a human value.

def derive_product_requirements_from_one_value(project_description: str, human_value: str) -> str:
    """
    Uses GPT-4o to derive product requirements given a human value.

    Args:
      human_value: A string describing a human value.

    Returns:
      A string containing the derived product requirements.
    """

    system_prompt = \
        """
        You are a helpful assistant that derives product requirements from human values.
        Provide the requirements in a clear and concise manner within a product development context.
        Use the following structure:
        1. **[Title]**
            - **Requirement:** [The directive of this requirement]
            - **Description:** [A brief description of the requirement, including any relevant details]
            - **Expected Product Behavior:** [Change in expected product behavior by adding this requirement]
            - **AI Risk Mitigation:** [The list of the AI Risks this requirement could help mitigate]
            - **Priority:** [Priority (low/medium/high)]
        """

    user_prompt = \
        f"""
        This is relevant information about the product I am working on:
        {project_description}

        Please derive product requirements from the following human value:
        {human_value}
        """

    # Remove indentation from the prompts
    system_prompt = inspect.cleandoc(system_prompt)
    user_prompt = inspect.cleandoc(user_prompt)

    client = openai.OpenAI(api_key=OPEN_AI_API_TOKEN)

    response = client.chat.completions.create(
                  model="gpt-4o-2024-08-06",
                  messages=[
                      {"role": "system", "content": system_prompt},
                      {"role": "user", "content": user_prompt}
                  ],
                  temperature=0,
                  max_tokens= 1024,
                  frequency_penalty=0.01,
               )

    return response.choices[0].message.content

def get_all_product_requirements(project_description: str, human_values: Dict[str, List[str]]) -> Dict[str, Dict[str, str]]:
    """
    Uses GPT-4o to derive product requirements given a list of human values.

    Args:
      human_values: A dictionary of strings describing human values.

    Returns:
      A string containing the derived product requirements.
    """

    human_value_groups = list(human_values.keys())
    product_requirements = {}
    for value_group in human_value_groups:
        product_requirements[value_group] = {}
        # product_requirements_for_value = {}
        for human_value in human_values[value_group]:
            product_requirements_for_value = derive_product_requirements_from_one_value(project_description, human_value)
            product_requirements[value_group][human_value] = product_requirements_for_value
    return product_requirements

def print_all_product_requirements(product_requirements: Dict[str, Dict[str, str]]):
    """
    Prints the derived product requirements.

    Args:
      product_requirements: A dictionary containing the derived product requirements.
    """

    for human_value_group in product_requirements.keys():
        print(f"# {human_value_group} Requirements")
        for human_value in product_requirements[human_value_group].keys():
            print(f"\n## Requirements for {human_value}\n")
            print(product_requirements[human_value_group][human_value])
        print("\n---\n")
```

### Example Usage

```python
product_requirements = get_all_product_requirements(organization_context, human_values)
print_all_product_requirements(product_requirements)
```
